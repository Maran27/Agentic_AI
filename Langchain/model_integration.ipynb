## Streaming and Batch

```python
# Now instead of waiting for the model to respond with the full response, we can use the `stream` method to get the response immediately. 
# That is the response is printed once the model start to generate the response that needs to be displayed. 
# This is useful for interactive applications where we want to get the response as soon as possible.

model.stream("What is the least common multiple of 12 and 18?") #only generator object not text

#for text
for chunk in model.stream("What is the least common multiple of 12 and 18?"):
    print(chunk.text, end="|", flush=True) #flush is used to print the text immediately without waiting for the buffer to fill up.

    
```

```python
#Batch processing is a way to process multiple inputs at once, instead of processing them one by one. 
#Nothing but parallel processing of multiple input requests to the LLM.
# This can be useful for improving performance and reducing latency when working with large datasets or when making multiple requests to a model.

responses = model.batch(
[
    "What is the least common multiple of 12 and 18?",
    "What is the greatest common divisor of 12 and 18?",
    "what is quantum computing?"
]
)
for response in responses:
    print(response)
    print(response.content, end="|", flush=True) #flush is used to print the text immediately without waiting for the buffer to fill up.

# we have max_concurrency parameter in the batch method which is used to limit the number of concurrent requests to the model.


responses = model.batch(
[
    "What is the least common multiple of 12 and 18?",
    "What is the greatest common divisor of 12 and 18?",
    "what is quantum computing?"
],
config={
    "max_concurrency": 5,
}
)
for response in responses:
    print(response)
```
