```python
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4.1")
model
```

```python
model.invoke("What is the capital of France?") # By default this is a Human Message, but you can also pass in a SystemMessage or AIMessage if you want to change the role of the message.
# But the output we receive back from this is a ChatResponse object, which contains the text of the response, as well as any tool calls that were made during the generation of the response. You can access the text of the response using the `text` attribute, and you can access the tool calls using the `tool_calls` attribute.
# But the output we receive back form this fucntion will be a AI Message
```

## Text Prompts

```python
model.invoke('what is AI') # no specific rules on how the model should behave. So these are called Text Prompts.
# Simply we give input human message and receive output message from the model as AI Message. 
```

## Message Prompts

```python
# Set of Instructions or messages are given to the model as prompts. These prompts can be of different types like HumanMessage, SystemMessage, AIMessage, ToolMessage. The model will respond based on the type of message it receives.
from langchain.messages import HumanMessage, SystemMessage, AIMessage, ToolMessage
messagess = [
    SystemMessage("You are a quantum physicist."), # tells about the initial role, tone, behavior of the model
    HumanMessage("What is Quantum entagenglement?"), # human message to the model
]
response = model.invoke(messagess) # model will respond based on the initial role and human message
```

```python
system_msg = SystemMessage("""
You are a senior python developer with expertise in qunatum computing.
Always provide code examples in your responses and explain the code in detail.
Be concise and clear in your explanations, avoiding unnecessary jargon.
""")
messages = [
    system_msg,
    HumanMessage("Can you provide a simple example of quantum entanglement in Python?")
]
response = model.invoke(messages)
print(response.context)  # This will print the model's response to the human message, which should include a code example and explanation.
```

```python
human_msg = HumanMessage(
    content="Hello",
    name="maran",
    id="12345"
)
response = model.invoke([human_msg])
response
```

```python
ai_msg = AIMessage("Hello! How can I assist you today?")
messages = [
    SystemMessage("You are a helpful assistant."),
    HumanMessage("Can you explain the concept of superposition in quantum mechanics?"),
    ai_msg,
    HumanMessage("Can you provide a simple example of superposition in Python?")
]
response = model.invoke(messages)
print(response.context)
print(response.usage_metadata)
```
